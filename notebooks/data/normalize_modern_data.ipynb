```python

import pandas as pd
from pathlib import Path

# Adjust path to your project's data directory
modern_df = pd.read_csv(
    "../../data/processed/modern_dataset_drift_eval.csv"
)

# Rename columns to match HC3
modern_df = modern_df.rename(
    columns={
        "answer": "text",
        "question_id": "id",
        "source": "question_source"
    }
)

# Preserve original generator identity
modern_df["generator"] = modern_df["model"]

# Match HC3 label conventions
modern_df["label"] = modern_df["label"].map({
    "human": "human",
    "ai": "chatgpt"
})

# Numeric labels: human = 0, AI = 1
modern_df["target"] = modern_df["label"].map({
    "human": 0,
    "chatgpt": 1
})

assert modern_df["text"].notna().all()
assert modern_df["target"].notna().all()
assert modern_df["id"].nunique() == 150
assert len(modern_df) == 600

print(modern_df["target"].value_counts())
print(pd.crosstab(
    modern_df["generator"],
    modern_df["target"]
))
```
